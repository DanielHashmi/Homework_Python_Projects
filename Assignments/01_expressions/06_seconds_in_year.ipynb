```python
def main():
    DAYS_PER_YEAR: int = 365
    HOURS_PER_DAY: int = 24
    MIN_PER_HOUR: int = 60
    SEC_PER_MIN: int = 60

    nunber_of_years_to_convert_to_seconds = int(input("Enter numbers of years to convert: "))
    number_of_leap_years = nunber_of_years_to_convert_to_seconds // 4  # leap year

    number_of_seconds_in_years = SEC_PER_MIN * MIN_PER_HOUR * HOURS_PER_DAY * (DAYS_PER_YEAR * nunber_of_years_to_convert_to_seconds + number_of_leap_years)
    print(f"There are {number_of_seconds_in_years} seconds in {nunber_of_years_to_convert_to_seconds} years")
if __name__ == '__main__':
    main()
```
