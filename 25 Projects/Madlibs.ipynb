```python
def madlibs():
  adjective1 = input("Enter an adjective: ")
  noun1 = input("Enter a noun: ")
  verb1 = input("Enter a verb (past tense): ")
  adjective2 = input("Enter another adjective: ")
  noun2 = input("Enter another noun: ")
  verb2 = input("Enter another verb: ")

  story = f"It was a {adjective1} day when I saw a {noun1} {verb1} down the street. \
It was so {adjective2} that I decided to follow the {noun2}. \
We eventually {verb2} together and became the best of friends."
  print(story)


if __name__ == "__main__":
  madlibs()
```
