```python
def main():
    countries = {
        '16': 'Peturksbouipo',
        '25': 'Stanlau',
        '48': 'Mayengua'
    }

    voter_age = input("How old are you? ")
    cannot_vote_ages = [age for age in countries if age != voter_age]

    if voter_age in countries:
        print(f'You can vote in {countries[voter_age]} where the voting age is {voter_age}. '
              f'You cannot vote in {countries[cannot_vote_ages[0]]} where the voting age is {cannot_vote_ages[0]}. '
              f'You cannot vote in {countries[cannot_vote_ages[1]]} where the voting age is {cannot_vote_ages[1]}.')
    else:
        print('Age out of range!')
if __name__ == '__main__':
    main()
```
