```python

def print_ones_digit(num):
    try:
        num = int(num)
        print("The once digit is:", num % 10)
    except ValueError:
        print("You entered a non numeric value!")

def main():
    num = input("Enter a number: ")
    print_ones_digit(num)
if __name__ == '__main__':
    main()
```
