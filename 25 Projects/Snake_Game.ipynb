```python
import pygame
import random

class Snake:
    def __init__(self, block_size):
        self.block_size = block_size
        self.reset()

    def reset(self):
        self.positions = [[250, 250]]
        self.direction = (0, 0)
        self.length = 1

    def move(self):
        head_x, head_y = self.positions[-1]
        delta_x, delta_y = self.direction
        new_head = [head_x + delta_x, head_y + delta_y]
        self.positions.append(new_head)
        if len(self.positions) > self.length:
            self.positions.pop(0)

    def draw(self, surface, color):
        for pos in self.positions:
            pygame.draw.rect(surface, color, (pos[0], pos[1], self.block_size, self.block_size))

class Food:
    def __init__(self, display_width, display_height, block_size):
        self.display_width = display_width
        self.display_height = display_height
        self.block_size = block_size
        self.position = self.random_position()

    def random_position(self):
        x = round(random.randrange(0, self.display_width - self.block_size) / 10.0) * 10.0
        y = round(random.randrange(0, self.display_height - self.block_size) / 10.0) * 10.0
        return [x, y]

    def draw(self, surface, color):
        pygame.draw.rect(surface, color, (self.position[0], self.position[1], self.block_size, self.block_size))


class SnakeGame:
    def __init__(self, width=500, height=500, snake_block=10, snake_speed=10):
        pygame.init()
        self.width = width
        self.height = height
        self.snake_block = snake_block
        self.snake_speed = snake_speed
        self.display = pygame.display.set_mode((self.width, self.height))
        pygame.display.set_caption("Snakky.io")
        self.clock = pygame.time.Clock()
        self.font_style = pygame.font.SysFont(None, 30)
        self.reset()

    def reset(self):
        self.snake = Snake(self.snake_block)
        self.food = Food(self.width, self.height, self.snake_block)
        self.game_over = False
        self.game_close = False

    def message(self, msg, color):
        mesg = self.font_style.render(msg, True, color)
        self.display.blit(mesg, (self.width / 6, self.height / 3))

    def process_events(self):
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                self.game_over = True
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_LEFT:
                    self.snake.direction = (-self.snake_block, 0)
                elif event.key == pygame.K_RIGHT:
                    self.snake.direction = (self.snake_block, 0)
                elif event.key == pygame.K_UP:
                    self.snake.direction = (0, -self.snake_block)
                elif event.key == pygame.K_DOWN:
                    self.snake.direction = (0, self.snake_block)

    def update(self):
        head = self.snake.positions[-1]
        if head[0] >= self.width or head[0] < 0 or head[1] >= self.height or head[1] < 0:
            self.game_close = True

        self.snake.move()

        head = self.snake.positions[-1]
        if head in self.snake.positions[:-1]:
            self.game_close = True

        if head[0] == self.food.position[0] and head[1] == self.food.position[1]:
            self.food.position = self.food.random_position()
            self.snake.length += 1

    def draw(self):
        self.display.fill((0, 0, 0))
        self.food.draw(self.display, (213, 50, 80))
        self.snake.draw(self.display, (0, 255, 0))
        pygame.display.update()

    def run(self):
        while not self.game_over:
            while self.game_close:
                self.display.fill((0, 0, 0))
                self.message("You Lost! Press C-Play Again or Q-Quit", (213, 50, 80))
                pygame.display.update()
                for event in pygame.event.get():
                    if event.type == pygame.KEYDOWN:
                        if event.key == pygame.K_q:
                            self.game_over = True
                            self.game_close = False
                        if event.key == pygame.K_c:
                            self.reset()
            self.process_events()
            self.update()
            self.draw()
            self.clock.tick(self.snake_speed)
        pygame.quit()
        quit()


if __name__ == "__main__":
    game = SnakeGame()
    game.run()
```
