```python
import math
class Board:
    def __init__(self):
        self.grid = [[" " for _ in range(3)] for _ in range(3)]

    def update_position(self, row, col, symbol):
        if self.grid[row][col] == " ":
            self.grid[row][col] = symbol
            return True
        return False

    def check_winner(self, symbol):
        for row in self.grid:
            if all(cell == symbol for cell in row):
                return True
        for col in range(3):
            if all(self.grid[row][col] == symbol for row in range(3)):
                return True
        if all(self.grid[i][i] == symbol for i in range(3)):
            return True
        if all(self.grid[i][2 - i] == symbol for i in range(3)):
            return True
        return False

    def is_full(self):
        return all(cell != " " for row in self.grid for cell in row)

    def available_moves(self):
        moves = []
        for i in range(3):
            for j in range(3):
                if self.grid[i][j] == " ":
                    moves.append((i, j))
        return moves

    def display(self):
        for row in self.grid:
            print("|".join(row))
        print()


class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol

    def get_move(self, board):
        raise NotImplementedError


class HumanPlayer(Player):
    def get_move(self, board):
        while True:
            try:
                move = int(input("Enter a number in range (0-8): "))
                if move < 0 or move > 8:
                    print("Move out of range!")
                    continue
                row = move // 3
                col = move % 3
                if board.grid[row][col] != " ":
                    print("Cell already occupied!")
                    continue
                return row, col
            except ValueError:
                print("Invalid input, please enter an integer.")


class ComputerPlayer(Player):
    def __init__(self, name, symbol, opponent_symbol):
        super().__init__(name, symbol)
        self.opponent_symbol = opponent_symbol

    def get_move(self, board):
        best_score = -math.inf
        best_move = None
        for move in board.available_moves():
            row, col = move
            board.grid[row][col] = self.symbol
            score = self.minimax(board, is_maximizing=False)
            board.grid[row][col] = " "
            if score > best_score:
                best_score = score
                best_move = move
        return best_move

    def minimax(self, board, is_maximizing):
        if board.check_winner(self.symbol):
            return 1
        if board.check_winner(self.opponent_symbol):
            return -1
        if board.is_full():
            return 0

        if is_maximizing:
            best_score = -math.inf
            for move in board.available_moves():
                row, col = move
                board.grid[row][col] = self.symbol
                score = self.minimax(board, is_maximizing=False)
                board.grid[row][col] = " "
                best_score = max(best_score, score)
            return best_score
        else:
            best_score = math.inf
            for move in board.available_moves():
                row, col = move
                board.grid[row][col] = self.opponent_symbol
                score = self.minimax(board, is_maximizing=True)
                board.grid[row][col] = " "
                best_score = min(best_score, score)
            return best_score


class Game:
    def __init__(self):
        self.board = Board()
        self.human = HumanPlayer("Player", "X")
        self.computer = ComputerPlayer("Computer", "O", "X")
        self.current_turn = self.human

    def play(self):
        game_over = False
        while not game_over:
            self.board.display()
            if isinstance(self.current_turn, HumanPlayer):
                print(f"{self.current_turn.name}'s turn ({self.current_turn.symbol})")
            else:
                print(f"{self.current_turn.name} is thinking...")
            row, col = self.current_turn.get_move(self.board)
            if self.board.update_position(row, col, self.current_turn.symbol):
                if self.board.check_winner(self.current_turn.symbol):
                    self.board.display()
                    print(f"{self.current_turn.name} wins!")
                    game_over = True
                elif self.board.is_full():
                    self.board.display()
                    print("It's a tie!")
                    game_over = True
                else:
                    self.current_turn = self.human if self.current_turn == self.computer else self.computer
            else:
                print("Invalid move, try again.")


if __name__ == "__main__":
    game = Game()
    game.play()
```
