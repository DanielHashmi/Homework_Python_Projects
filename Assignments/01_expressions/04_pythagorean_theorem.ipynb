```python
import math
def main():
    A:float = float(input("Enter first side of the right triangle: "))
    B:float = float(input("Enter second side of the right triangle: "))
    C:float = math.sqrt(A**2 + B**2)
    print("The length of the hypotenuse is", C)

if __name__ == '__main__':
    main()
```
