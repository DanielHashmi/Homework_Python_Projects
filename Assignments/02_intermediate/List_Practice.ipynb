# List Practice

```python
fruits = ['apple', 'banana', 'orange', 'grape', 'pineapple']
```

```python
len(fruits)
```

```python
fruits.append('mango')
```

```python
fruits
```

```python
def get_element(lst, inx):
    try:
        return lst[inx]
    except IndexError:
        return "Element not found in list!"
```

```python
get_element(fruits,12)
```

```python
def update_ele(lst, inx, new_val):
    try:
        lst[inx] = new_val
        return lst
    except IndexError:
        return "Wrong Updation!"
```

```python
fruits
```

```python
update_ele(fruits, 3, 'kiwi')
```

```python
def slice_list(lst, start_i, end_i):
    try:
        return lst[start_i:end_i]
    except IndexError:
        return "Wrong Slicing!"
```

```python
slice_list(fruits,2,4)
```

```python
operations = {
    'access': get_element,
    'modify': update_ele,
    'slice': slice_list
}

def main():
    operation = input("Enter operation (access/modify/slice): ")
    if operation.lower() == 'access':
        lst = input("Enter list elements (comma-separated -> apple, banana): ").replace(' ','').split(',')
        inx = int(input("Enter index: "))
        print(operations[operation](lst, inx))
    elif operation.lower() == 'modify':
        lst = input("Enter list elements (comma-separated -> apple, banana): ").replace(' ','').split(',')
        inx = int(input("Enter index: "))
        new_val = input("Enter new value: ")
        print(operations[operation](lst, inx, new_val))
    else:
        lst = input("Enter list elements (comma-separated -> apple, banana): ").replace(' ','').split(',')
        start_i = int(input("Enter start index: "))
        end_i = int(input("Enter end index: "))
        print(operations[operation](lst, start_i, end_i))
if __name__ == '__main__':
    main()
```
