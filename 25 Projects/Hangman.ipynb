```python
import random as r

words = ['Banana', 'Apple', 'Bean', 'Kite']
word = r.choice(words).upper()
hidden = list('⁕' * len(word))
hearts = 5
words_used = []
while hearts > 0:
    print('💗 ' * hearts)
    print(' '.join(hidden))
    if ''.join(hidden) == word:
        print('You Survived!')
        break
    print(f"Words Used: {', '.join(words_used)}")
    l = input("Guess a Letter: ").upper()

    if l in words_used:
        print(f"You already guessed {l}!")
        continue
    words_used.append(l)

    if l in word:
        for i, letter in enumerate(word):
            if letter == l:
                hidden[i] = l
    else:
        hearts -= 1
        print('Incorrect Letter!')

else:
    print('You Died!')
```
