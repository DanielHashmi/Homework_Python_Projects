```python
def check_leap_year(year):
    try:
        y = int(year)
        if (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0):
            return "It's a leap year"
        return "It's not a leap year"
    except ValueError:
        return "Incorrect year!"

print(check_leap_year(input("Please input a year: ")))
```
