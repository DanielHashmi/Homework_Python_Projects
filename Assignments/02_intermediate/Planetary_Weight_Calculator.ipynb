```python
planet_weights = {
    "Mercury": 0.376,
    "Venus": 0.889,
    "Mars": 0.378,
    "Jupiter": 2.36,
    "Saturn": 1.081,
    "Uranus": 0.815,
    "Neptune": 1.14
}

def main():
    try:
        weight = int(input("Enter a weight on Earth: "))
        planet = input('Enter a planet: ').capitalize()
        print(f"The equivalent on {planet}: {round(weight* planet_weights[planet], 2)}")
    except (ValueError, KeyError):
        print("Incorrect Input!")
if __name__ == '__main__':
    main()
```
