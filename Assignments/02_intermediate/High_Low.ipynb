```python
import random

NUM_ROUNDS = 5
def generate_rand_n():
    return random.randint(0, 100)

def main():
    r = 1
    score = 0
    print("Welcome to the High-Low Game!")
    print('--------------------------------')
    while r <= NUM_ROUNDS:
        your_n = generate_rand_n()
        bot_n = generate_rand_n()
        print("Round", r)
        print('Your number is', your_n)
        inp = input("Do you think your number is higher or lower than the computer's?: ")
        while inp.lower() not in ['higher','lower']:
            inp = input("Please enter either higher or lower: ")

        if (inp.lower() == 'higher' and your_n > bot_n) or (inp.lower() == 'lower' and your_n < bot_n):
            print("You were right! The computer's number was", bot_n)
            score += 1
            print('Your score is now', score)
        else:
            print("Aww, that's incorrect. The computer's number was", bot_n)
            print('Your score is now', score)
        r+=1

    if score == NUM_ROUNDS:
        print('Wow! You played perfectly!')
    elif score > NUM_ROUNDS // 2:
        print('Good job, you played really well!')
    else:
        print('Better luck next time!')

    print("Thanks for playing!")
if __name__ == '__main__':
    main()
```
