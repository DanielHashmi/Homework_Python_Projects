```python
fruits = {
    "mango":20,
    "apple":10,
    "banana":6,
    "strawberry":12,
    "kiwi":9,
    "orange":11
}

totel_price = 0
for fruit in fruits:
    user = int(input(f"How many {fruit} do you want to buy? "))
    totel_price += fruits[fruit] * user

print(f"Totel: {totel_price}")
```
