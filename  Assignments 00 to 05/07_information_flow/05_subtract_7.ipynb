```python
def subtract_seven(num):
	num -= 7
	return num

def main():
    try:
        num = int(input("Enter a number to subtract 7 from it: "))
        print(f"Subtracting (7) from number ({num}) is ({subtract_seven(num)})")
    except ValueError:
        print("Invalid Number!")

if __name__ == '__main__':
    main()
```
