```python
import random

def main():
    range = 10
    computer = random.randint(0,range)
    guess = None
    turns = 0
    while guess != computer:
        try:
            guess = int(input(f"Guess a number in range (0-{range}): "))
            turns += 1
            if guess > computer:
                print("Too high!")
            elif guess < computer:
                print("Too low!")
        except ValueError:
            print(f"Incorrect number! Please enter in range (0-{range})")
    print(f"Great! you guessed the number: {guess} in {turns} turns.")
if __name__ == '__main__':
    main()
```
