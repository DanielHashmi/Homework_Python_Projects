```python
class Board:
    def __init__(self):
        self.grid = [[" " for _ in range(3)] for _ in range(3)]

    def update_position(self, row, col, symbol):
        if self.grid[row][col] == " ":
            self.grid[row][col] = symbol
            return True
        return False

    def check_winner(self, symbol):
        win_patterns = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],
            [0, 3, 6], [1, 4, 7], [2, 5, 8],
            [0, 4, 8], [2, 4, 6]
        ]
        return any(all(self.grid[i // 3][i % 3] == symbol for i in pattern) for pattern in win_patterns)

    def display(self):
        for row in self.grid:
            print("|".join(row))
            print("-" * 5)


class Player:
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol

    def get_move(self, board):
        while True:
            try:
                player = int(input("Enter a number in the range of (0-8): "))
                if player < 0 or player > 8:
                    print("Input out of range. Try again.")
                    continue
                row = player // 3
                col = player % 3
                if board.grid[row][col] != " ":
                    print("Cell already occupied. Try again.")
                    continue
                return (row, col)
            except ValueError:
                print("Incorrect input! Enter an integer between 0–8.")


class Game:
    def __init__(self):
        self.board = Board()
        self.players = [
            Player("Player X", "X"),
            Player("Player O", "O")
        ]
        self.current_turn = 0

    def play(self):
        game_over = False
        while not game_over:
            current_player = self.players[self.current_turn]
            print(f"{current_player.name}'s turn ({current_player.symbol})")
            row, col = current_player.get_move(self.board)
            if self.board.update_position(row, col, current_player.symbol):
                self.board.display()
                if self.board.check_winner(current_player.symbol):
                    print(f"{current_player.name} wins!")
                    game_over = True
                else:
                    if all(cell != " " for row in self.board.grid for cell in row):
                        print("It's a tie!")
                        game_over = True
                    else:
                        self.current_turn = (self.current_turn + 1) % len(self.players)
            else:
                print("Invalid move, try again.")


if __name__ == "__main__":
    game = Game()
    game.play()
```
