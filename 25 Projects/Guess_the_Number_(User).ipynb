```python
import random

low = 0
high = 10
user = 6

while low <= high:
    com = random.randint(low, high)
    if user != com:
        while True:
            hint = input(f"Is {com} your number: (H) for high, (L) for low!").lower()
            if hint in ('h', 'l'):
                break
            else:
                print("Invalid input. Please enter 'H' or 'L'.")
        if hint == 'h':
            low = com + 1
        elif hint == 'l':
            high = com - 1
    else:
        print(f"Yay! computer guessed your number: {com}")
        break
```
