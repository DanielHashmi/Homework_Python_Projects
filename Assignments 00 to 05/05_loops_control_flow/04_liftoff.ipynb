```python
import time
for i in range(1,11):
    print(11 - i)
    time.sleep(0.3)

print('Liftoff!')

number = int(input("Enter a number: "))
curr_value = number
final_value = 100

while curr_value < final_value:
    curr_value *= 2
    print(curr_value)
```
