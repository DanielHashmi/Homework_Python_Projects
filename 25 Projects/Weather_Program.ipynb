```python
import requests
from datetime import datetime

def get_weather(location, api_key):
    base_url = "https://api.openweathermap.org/data/2.5/weather"
    params = {
        'q': location,
        'appid': api_key,
        'units': 'metric'
    }

    try:
        response = requests.get(base_url, params=params)
        response.raise_for_status()
        data = response.json()

        return {
            'city': data['name'],
            'country': data['sys']['country'],
            'temp': data['main']['temp'],
            'feels_like': data['main']['feels_like'],
            'description': data['weather'][0]['description'],
            'humidity': data['main']['humidity'],
            'wind_speed': data['wind']['speed'],
            'datetime': datetime.fromtimestamp(data['dt']).strftime('%Y-%m-%d %H:%M:%S')
        }
    except Exception as e:
        return f"An error occurred: {str(e)}"

def display_weather(weather_data):
    if isinstance(weather_data, str):
        print(weather_data)
        return

    print("\n===== Weather Information =====")
    print(f"Location: {weather_data['city']}, {weather_data['country']}")
    print(f"Date/Time: {weather_data['datetime']}")
    print(f"Temperature: {weather_data['temp']}°C (Feels like: {weather_data['feels_like']}°C)")
    print(f"Condition: {weather_data['description'].capitalize()}")
    print(f"Humidity: {weather_data['humidity']}%")
    print(f"Wind Speed: {weather_data['wind_speed']} m/s")
    print("==============================\n")

def main():
    print("===== Weather Information Program =====")
    print("Get real-time weather data for any city!")

    api_key = input("\nPlease enter your OpenWeatherMap API key: ")

    while True:
        location = input("\nEnter city name (or 'exit' to quit): ")

        if location.lower() == 'exit':
            print("Thank you for using the Weather Program. Goodbye!")
            break
        weather_data = get_weather(location, api_key)
        display_weather(weather_data)

if __name__ == "__main__":
    main()
```
