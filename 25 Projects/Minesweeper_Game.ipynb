```python
import random

class Minesweeper:
    def __init__(self, rows, cols, num_mines):
        self.rows = rows
        self.cols = cols
        self.num_mines = num_mines
        self.board = [[' ' for _ in range(cols)] for _ in range(rows)]
        self.mine_locations = []
        self.revealed_cells = set()

    def generate_mines(self):
        while len(self.mine_locations) < self.num_mines:
            row = random.randint(0, self.rows - 1)
            col = random.randint(0, self.cols - 1)
            if (row, col) not in self.mine_locations:
                self.mine_locations.append((row, col))

    def count_adjacent_mines(self, row, col):
        count = 0
        for i in range(max(0, row - 1), min(self.rows, row + 2)):
            for j in range(max(0, col - 1), min(self.cols, col + 2)):
                if (i, j) in self.mine_locations:
                    count += 1
        return count

    def reveal_cell(self, row, col):
        if (row, col) in self.revealed_cells or row < 0 or row >= self.rows or col < 0 or col >= self.cols:
            return

        self.revealed_cells.add((row, col))

        if (row, col) in self.mine_locations:
            return True

        adjacent_mines = self.count_adjacent_mines(row, col)
        if adjacent_mines > 0:
            self.board[row][col] = str(adjacent_mines)
        else:
            self.board[row][col] = ' '
            for i in range(max(0, row - 1), min(self.rows, row + 2)):
                for j in range(max(0, col - 1), min(self.cols, col + 2)):
                    self.reveal_cell(i, j)

        return False

    def print_board(self):
        print("   " + " ".join(str(i) for i in range(self.cols)))
        for i in range(self.rows):
            row_str = str(i) + "  "
            for j in range(self.cols):
                if (i, j) in self.revealed_cells:
                    if (i, j) in self.mine_locations:
                        row_str += "* "
                    else:
                        adjacent_mines = self.count_adjacent_mines(i, j)
                        row_str += (str(adjacent_mines) if adjacent_mines > 0 else ' ') + " "
                else:
                    row_str += ". "
            print(row_str)

    def play(self):
        self.generate_mines()
        while True:
            self.print_board()
            try:
                row = int(input("Enter row: "))
                col = int(input("Enter column: "))

                if row < 0 or row >= self.rows or col < 0 or col >= self.cols:
                    print("Invalid input. Please enter valid row and column.")
                    continue

                game_over = self.reveal_cell(row, col)
                if game_over:
                    self.print_board()
                    print("Game Over! You hit a mine.")
                    break

                if len(self.revealed_cells) == self.rows * self.cols - self.num_mines:
                    self.print_board()
                    print("Congratulations! You win!")
                    break

            except ValueError:
                print("Invalid input. Please enter numbers only.")


if __name__ == "__main__":
    rows = 9
    cols = 9
    num_mines = 10
    game = Minesweeper(rows, cols, num_mines)
    game.play()
```
