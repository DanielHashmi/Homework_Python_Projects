```python
import requests
from bs4 import BeautifulSoup
import re
import sys

def get_github_profile_image(github_username_or_url):
    if "github.com" in github_username_or_url:
        match = re.search(r'github\.com/([^/]+)', github_username_or_url)
        if match:
            username = match.group(1)
        else:
            return "Invalid GitHub URL format"
    else:
        username = github_username_or_url

    url = f"https://github.com/{username}"
    try:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
        }
        response = requests.get(url, headers=headers, timeout=10)

        if response.status_code == 200:
            soup = BeautifulSoup(response.text, 'html.parser')
            img_tag = soup.select_one('img.avatar.avatar-user')
            if img_tag and img_tag.has_attr('src'):
                return img_tag['src']
        return f"Failed to fetch profile page: HTTP {response.status_code}"
    except requests.exceptions.RequestException as e:
        return f"Error during request: {e}"

def main():
    github_input = input("Enter GitHub username or profile URL: ").strip()
    if not github_input:
        print("No input provided. Exiting.")
        sys.exit(1)
    profile_image_url = get_github_profile_image(github_input)
    print(f"\nProfile Image URL: {profile_image_url}")

if __name__ == "__main__":
    main()
```
