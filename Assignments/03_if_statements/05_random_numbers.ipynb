```python
import random as r

N_NUMBERS: int = 10
MIN_VALUE: int = 1
MAX_VALUE: int = 100

def get_random_number():
    return r.randint(MIN_VALUE, MAX_VALUE)

def main():
    for _ in range(0, N_NUMBERS):
        print(get_random_number())

if __name__ == '__main__':
    main()
```
