```python
def main():
    M = float(input("Enter kilos of mass: "))
    C = 299792458
    E = M * C**2
    print("e = m * C^2...")
    print(f"m = {M} kg")
    print(f"C = {C} m/s")
    print(f"{E} joules of energy!")

if __name__ == '__main__':
    main()
```
