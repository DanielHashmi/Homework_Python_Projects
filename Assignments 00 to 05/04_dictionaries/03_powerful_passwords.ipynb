```python
from hashlib import sha256

def login(email, database, password):
    if email in database and database[email] == hash(password):
        return "Logged In!"
    return "Invalid Credentials!"

def hash(password):
    return sha256(password.encode()).hexdigest()

def main():
    database = {}

    while True:
        email = input("Enter your email or type 'exit' to end: ").strip()
        if email.lower() == 'exit':
            break

        password = input("Enter your password or type 'exit' to end: ").strip()
        if password.lower() == 'exit':
            break

        database[email] = hash(password)
        print("Credentials saved! Add more or type 'exit' to finish.")

    if database:
        ask = input("Do you want to login? (yes/no): ").strip().lower()
        if ask == 'yes':
            email = input("Enter your email: ").strip()
            password = input("Enter your password: ").strip()
            print(login(email, database, password))
    else:
        print("No credentials were stored. Exiting.")

if __name__ == '__main__':
    main()
```
