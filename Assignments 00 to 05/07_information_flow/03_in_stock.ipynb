```python
inventory = {
    "apple": 39,
    "banana": 343,
    "orange": 23,
    "kiwi": 234,
    "mango": 399,
    "strawberry": 324,
    "grapefruit": 33
}


def num_in_stock(fruit):
    return inventory[fruit]


def main():
    fruit = input("Enter a fruit: ")
    try:
        print(f"This fruit is in stock! Here is how many:\n{inventory[fruit]}")
    except KeyError:
        print("This fruit is not in stock.")


if __name__ == '__main__':
    main()
```
