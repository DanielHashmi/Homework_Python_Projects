```python
MINIMUM_HEIGHT : int = 50

def main():
    try:
        height = float(input("How tall are you? "))
        while height != '':
            if height >= MINIMUM_HEIGHT:
                print("You're tall enough to ride!")
            else:
                print("You're not tall enough to ride, but maybe next year!")
            height = float(input("How tall are you? "))
    except ValueError:
        pass

if __name__ == '__main__':
    main()
```
