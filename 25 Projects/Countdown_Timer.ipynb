```python
import time
def start_timer(count_down):
    while count_down:
        mins, secs = divmod(count_down, 60)
        print(f"{mins:02d}:{secs:02d}")
        count_down -= 1
        time.sleep(1)
    print('Time is up!')

while True:
    try:
        count_down = int(input("Enter Count Down: "))
        break
    except ValueError:
        print("Invalid input. Please enter a number.")

start_timer(count_down)
```
