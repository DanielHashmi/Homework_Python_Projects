```python
import random

N_NUMBERS: int = 10
MIN_VALUE: int = 1
MAX_VALUE: int = 100

def get_random_number():
    return random.randint(MIN_VALUE, MAX_VALUE)

def main():
  for _ in range(10):
      print(get_random_number(),end=' ')

if __name__ == '__main__':
    main()
```
