```python
d = {}
while True:
    user = input("Enter a number: ")
    if not user:
        break
    if not user in d:
        d.update({user: 1})
        continue
    d.update({user:d[user]+1})

for key, value in d.items():
    print(f"{key} appears {value} times.")
```
